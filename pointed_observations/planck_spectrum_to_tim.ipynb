```python

def ra_to_deg(ra_string):
    # Example input: "07h 22m 58.33s"
    h, m, s = (
    ra_string
    .replace("h", "")
    .replace("m", "")
    .replace("s", "")
    .split()
    )
    return (float(h) + float(m)/60 + float(s)/3600) * 1

def dec_to_deg(dec_string):
    # Example input: "-25d 46' 02.92''"
    dec_string = (
    dec_string
    .replace("d", " ")
    .replace("'", " ")
    .replace('"', " ")
    )
    parts = dec_string.split()
    sign = -1 if parts[0].startswith("-") else 1
    deg = abs(float(parts[0]))
    arcmin = float(parts[1])
    arcsec = float(parts[2])
    return sign * (deg + arcmin/60 + arcsec/3600)
```
